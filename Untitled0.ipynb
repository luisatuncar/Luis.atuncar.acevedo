```python
import numpy as np
import matplotlib.pyplot as plt

# =========================================
# DATOS
# =========================================

f = np.array([
    100,120,145,170,200,235,270,310,355,405,
    460,520,585,655,730,810,895,985,1080,1180,
    1290,1410,1540,1680,1830,1990,2160,2340,2530,2730
])

Z = np.array([
    152.3,149.1,146.8,144.9,142.0,139.5,137.9,136.1,134.8,133.6,
    132.7,131.9,131.4,131.1,130.9,131.0,131.3,131.9,132.7,133.8,
    135.2,136.9,138.9,141.1,143.5,146.1,149.0,152.2,155.6,159.2
])

# =========================================
# BUSCAR MÍNIMO LOCAL
# =========================================

indice_min = np.argmin(Z)

f_min = f[indice_min]
Z_min = Z[indice_min]

# =========================================
# MOSTRAR RESULTADOS
# =========================================

print("===== MÍNIMO LOCAL =====\n")

print(f"Frecuencia aproximada = {f_min} Hz")
print(f"|Z|min = {Z_min} ohm")

# =========================================
# GRÁFICO
# =========================================

plt.figure(figsize=(10,6))

plt.plot(f, Z,
         marker='o',
         color='blue',
         linewidth=2,
         label='|Z|(f)')

# marcar mínimo
plt.scatter(f_min, Z_min,
            color='red',
            s=100,
            label='Mínimo local')

plt.axvline(f_min,
            color='red',
            linestyle='--')

plt.xlabel('Frecuencia (Hz)')
plt.ylabel('|Z| (ohm)')

plt.title('Identificación del mínimo local')

plt.legend()
plt.grid()

plt.show()

# =========================================
# DISCUSIÓN
# =========================================

print("\n===== DISCUSIÓN =====\n")

print("Visualmente se observa un mínimo local")
print("en la curva de impedancia.")

print(f"El mínimo ocurre aproximadamente en {f_min} Hz,")
print(f"con un valor de |Z| ≈ {Z_min} ohm.")

print("La curva disminuye inicialmente y luego aumenta,")
print("indicando un comportamiento típico de sistemas")
print("con componentes resistivos y reactivos.")
```
